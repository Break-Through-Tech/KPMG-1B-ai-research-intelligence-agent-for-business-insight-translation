```python
import arxiv

search = arxiv.Search(
    query="cat:cs.AI",
    max_results=10,
    sort_by=arxiv.SortCriterion.SubmittedDate
)

client = arxiv.Client()
papers = list(client.results(search))

for paper in papers:
    print(paper.title)
    print(paper.published)
    print(paper.entry_id)
    print("-" * 50)
```

```python
paper = papers[0]

print("Title:", paper.title)
print("Authors:", [author.name for author in paper.authors])
print("Published:", paper.published)
print("Categories:", paper.categories)
print("PDF URL:", paper.pdf_url)
print("Entry ID:", paper.entry_id)
print("\nAbstract:\n", paper.summary)
```

```python
import pandas as pd

data = []

for paper in papers:
    data.append({
        "title": paper.title,
        "authors": [author.name for author in paper.authors],
        "published": paper.published,
        "categories": paper.categories,
        "pdf_url": paper.pdf_url,
        "entry_id": paper.entry_id,
        "abstract": paper.summary
    })

df = pd.DataFrame(data)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["abstract_length"] = df["abstract"].str.len()

df[["title", "abstract_length"]]
```

```python
df["abstract_length"].describe()
```

```python
from collections import Counter

category_counts = Counter(
    category
    for categories in df["categories"]
    for category in categories
)

category_counts
```

The initial sample contains complete metadata with no missing values. Abstracts average about 1,431 characters. All papers are categorized as cs.AI, while some also belong to additional categories such as computer vision, human-computer interaction, machine learning, and robotics. This suggests that category filtering alone may not be enough to determine business relevance.

## Larger Sample Exploration

```python
search_100 = arxiv.Search(
    query="cat:cs.AI",
    max_results=100,
    sort_by=arxiv.SortCriterion.SubmittedDate
)

papers_100 = list(client.results(search_100))

data_100 = []

for paper in papers_100:
    data_100.append({
        "title": paper.title,
        "authors": [author.name for author in paper.authors],
        "published": paper.published,
        "categories": paper.categories,
        "pdf_url": paper.pdf_url,
        "entry_id": paper.entry_id,
        "abstract": paper.summary
    })

df_100 = pd.DataFrame(data_100)

df_100.head()
```

```python
df_100.info()
```

```python
df_100.isnull().sum()
```

```python
print("Duplicate IDs:", df_100["entry_id"].duplicated().sum())
print("Duplicate titles:", df_100["title"].duplicated().sum())
```

```python
df_100["abstract_length"] = df_100["abstract"].str.len()
df_100["abstract_length"].describe()
```

```python
df_100["author_count"] = df_100["authors"].apply(len)
df_100["author_count"].describe()
```

```python
from collections import Counter

category_counts_100 = Counter(
    category
    for categories in df_100["categories"]
    for category in categories
)

category_counts_100.most_common(15)
```

```python
print("Earliest:", df_100["published"].min())
print("Latest:", df_100["published"].max())
```

### Larger Sample Findings

The 100-paper sample had no missing values or duplicate papers. Abstract length and author count varied across papers, and many papers belonged to multiple AI-related categories. This shows that category filtering alone may not be enough to determine relevance.

The sample also covers only a few hours from one day, so a larger dataset across a wider date range may be needed. Before building the retrieval pipeline, the data should be cleaned, normalized, and prepared for text chunking and embeddings.
